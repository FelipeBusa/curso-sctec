**f-string** = Formatação texto print()

````python
nome = "Mariana"
idade = 28
print(f"Olá, {nome}! Você tem {idade} anos.")
````

---
**\n** = \n no texto faz pular uma linha no print

**:.2f** = No f-string serve para mostrar 2 casas após a virgula

````python
dinheiro = 10
print(f"Eu tenho\nR$ {dinheiro:.2f} reais")
````

| Símbolo | Nome | Usos em Python | Exemplos |

### 5. Estruturas de Repetição (for / while)
***for*** = quando você já sabe quantas vezes quer repetir.

***while*** = quando a repetição depende de uma condição que pode mudar.

````python
# Exemplo 'for'
for numero in range(1,5):
    if numero %2 == 0:
        print(numero)
````

````python
# Exemplo 'while'
contador = 2
while contador > 0:
    print(contador)
    contador -= 1
````
