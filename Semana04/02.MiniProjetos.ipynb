## Desafio 1 - Dashboard RH
Arquivo: cadastro_funcionario.csv

Gerar um relatório contendo:
* Total de funcionários
* Funcionários por setor
* Funcionários por cargo
* Ranking dos setores
* Ranking dos cargos
* Setor destaque
* Cargo destaque
* Salvar o resultado em: dashboard_rh.json

```python
import csv
import json

# =========================
# LEITURA DO CSV
with open ("dataset/cadastro_funcionario.csv", encoding="utf-8") as arquivo:
    leitor = csv.DictReader(arquivo)
    funcionarios = list(leitor)


# =========================
# TOTAL DE FUNCIONÁRIOS
total_funcionarios = len(funcionarios)



# =========================
# CONTAGEM POR SETOR
contagem_setores = {}

for f in funcionarios:
    setor = f["setor"]
    contagem_setores[setor] = contagem_setores.get(setor, 0) + 1


# =========================
# CONTAGEM POR CARGO
contagem_cargos = {}

for f in funcionarios:
    cargo = f["cargo"]
    contagem_cargos[cargo] = contagem_cargos.get(cargo, 0) + 1
  
    
# =========================
# RANKING DOS SETORES

# Ordem alfabética
ranking_setores_alfabetica = sorted(contagem_setores.items())
    
# Ordem crescente
ranking_setores_crescente = sorted(
    contagem_setores.items(),
    key=lambda x: x[1],
    reverse=True    
    )
    
    
# =========================
# RANKING DOS CARGOS

# Ordem alfabética
ranking_cargo_alfabetica = sorted(contagem_cargos.items())

# Ordem crescente
ranking_cargos_crescente = sorted(
    contagem_cargos.items(),
    key=lambda x: x[1],
    reverse=True    
    )

    
# =========================
# SETOR DESTAQUE
setor_destaque = ranking_setores_crescente[0]

# =========================
# CARGO DESTAQUE
cargo_destaque = ranking_cargos_crescente[0]


# =========================
# CRIANDO O DASHBOARD
dashboard_rh = {
    "total_funcionarios": total_funcionarios,

    "funcionarios_por_setor": contagem_setores,

    "funcionarios_por_cargo": contagem_cargos,

    "ranking_setores": [
        {
            "setor": setor,
            "quantidade": quantidade
        }
        for setor, quantidade in ranking_setores_crescente
    ],

    "ranking_cargos": [
        {
            "cargo": cargo,
            "quantidade": quantidade
        }
        for cargo, quantidade in ranking_cargos_crescente
    ],

    "setor_destaque": {
        "setor": setor_destaque[0],
        "quantidade": setor_destaque[1]
    },

    "cargo_destaque": {
        "cargo": cargo_destaque[0],
        "quantidade": cargo_destaque[1]
    }
}

# =========================
# EXIBINDO O RELATÓRIO
print("=" * 30)
print("DASHBOARD RH")
print("=" * 30)

print(f"Total de funcionários: {total_funcionarios}")

print("\nFUNCIONÁRIOS POR SETOR")
for setor, quantidade in contagem_setores.items():
    print(f"{setor} | {quantidade}")

print("\nFUNCIONÁRIOS POR CARGO")
for cargo, quantidade in contagem_cargos.items():
    print(f"{cargo} | {quantidade}")

print("\nRANKING DOS SETORES")
for setor, quantidade in ranking_setores_crescente:
    print(f"{setor} | {quantidade}")

print("\nRANKING DOS CARGOS")
for cargo, quantidade in ranking_cargos_crescente:
    print(f"{cargo} | {quantidade}")

print("\nSETOR DESTAQUE")
print(f"{setor_destaque[0]} | {setor_destaque[1]}")

print("\nCARGO DESTAQUE")
print(f"{cargo_destaque[0]} | {cargo_destaque[1]}")


# =========================
# SALVANDO EM JSON
with open("dataset/dashboard_rh.json", "w", encoding="utf-8") as arquivo:
    json.dump(
        dashboard_rh,
        arquivo,
        ensure_ascii=False,
        indent=4
    )


# =========================
# LENDO O JSON CRIADO
with open("dataset/dashboard_rh.json", encoding="utf-8") as arquivo:
    dashboard_lido = json.load(arquivo)

print("\nJSON criado com sucesso!")
```

## Desafio 2 - Sistema de Auditoria de E-mails
Utilizando: emails_corporativos.json

* Criar funções para:
    * validar e-mail
    * contar válidos
    * contar inválidos

* Listar inválidos
* Gerar relatório
* Depois salvar: relatorio_emails.json

```python
import json # ler arquivos json
import re # validar e-mails

# ====================
# LENDO ARQUIVO JSON
with open("dataset/emails_corporativos.json", encoding="utf-8") as arquivo:
    emails = json.load(arquivo)


# ====================
# CRIANDO DEF VALIDAR E-MAILS
def validar_email(email):
    padrao = r"^[^\s@]+@[^\s@]+\.[^\s@]+$"
    
    if re.match(padrao, email):
        return True
    else:
        return False


# ====================
# CRIANDO DEF LISTAGEM EMAILS INVALIDOS
def listar_invalidos(emails):
    lista_invalidos = []

    for item in emails:
        email = item["email"]

        if not validar_email(email):
            lista_invalidos.append(email)

    return lista_invalidos


# ====================
# CRIANDO DEF CONTAR E-MAILS VALIDOS
def contar_validos(emails):
    email_validos = 0
    
    for item in emails:
        email = item["email"]
        
        if validar_email(email):
            email_validos += 1
    
    return email_validos

# ====================
# CRIANDO DEF CONTAR E-MAIL INVALIDOS
def contar_invalidos(emails):
    email_invalidos = 0
    
    for item in emails:
        email = item["email"]
        
        if not validar_email(email):
            email_invalidos += 1
    
    return email_invalidos


# # ====================
# # LISTANDO VALIDACAO COMPLETA DOS EMAILS
# for email in emails:
#     email = email['email']
    
#     if validar_email(email):
#         print(f"E-mail: {email} | E-mail válido")
#     else:
#         print(f"E-mail: {email} | E-mail inválido")


# ====================
# CRIANDO RELATORIO DE EMAILS
relatorio = {
    "total_emails": len(emails),
    "email_validos": contar_validos(emails),
    "email_invalidos": contar_invalidos(emails),
    "lista_emails_invalidos": listar_invalidos(emails)
}



# ====================
# VISUALIZAÇÃO DO RELATORIO
print("=" * 40)
print("RELATÓRIO DE E-MAILS")
print("=" * 40)

print(f"Total de e-mails: {relatorio['total_emails']}")
print(f"Total de e-mail VÁLIDOS: {relatorio['email_validos']}")
print(f"Total de e-mails INVÁLIDOS: {relatorio['email_invalidos']}")

print("")
print("lista de e-mails INVÁLIDOS:")

for item in relatorio['lista_emails_invalidos']:
    print(f"- {item}")
    

# # ====================
# # CRIANDO ARQUIVO
# with open("dataset/relatorio_emails.json", "w", encoding="utf-8") as arquivo:
#     json.dump(relatorio, arquivo, ensure_ascii=False, indent=4)
    
    
# ====================
# VALIDANDO CRIAÇÃO
with open("dataset/relatorio_emails.json", encoding="utf-8") as arquivo:
    validacao_relatorio_json = json.load(arquivo)

print()
print("=" * 40)
print("VALIDAÇÃO RELATÓRIO JSON")
print("=" * 40)
print(validacao_relatorio_json)
```
