## Desafio 1 - Cadastro de Funcionários
Arquivo: cadastro_funcionario.csv

Faça:
* Ler o CSV.
* Contar funcionários por setor.
* Contar funcionários por cargo.
* Descobrir qual setor possui mais funcionários.
* Descobrir qual cargo possui menos funcionários.

Objetivo - Treinar:
* CSV
* lista de dicionários
* criação de dicionários resumo

```python
# importar leitor csv
import csv

# abrir e ler arquivo csv
with open("dataset/cadastro_funcionario.csv", encoding="utf-8") as arquivo:
    leitor = csv.DictReader(arquivo)
    funcionarios = list(leitor)

# # Print iniciais para mostrar resultados básicos
# print(type(funcionarios))
# print(funcionarios[0])
# print(funcionarios[0].keys())

# =============================
# 1. CADASTRO DE FUNCIONÁRIOS
print("=" * 40)
print("1. CADASTRO DE FUNCIONÁRIOS")
print("=" * 40)

# Quantidade funcionários
print(f"Total de funcionários: {len(funcionarios)}")


# =============================
# 2. CALCULO DE FUNCIONÁRIOS
print("")
print("=" * 40)
print("2. CALCULO DE FUNCIONÁRIOS")
print("=" * 40)

# calculo quantidade funcionários:
setores = {}

for funcionario in funcionarios:
    setor = funcionario["setor"]

    if setor not in setores:
        setores[setor] = 1

    else:
        setores[setor] += 1

for setor, quantidade in setores.items():
    print(f"{setor} ..... {quantidade} funcionário(s)")



# calculo setor com mais e menor funcionários
primeiro_setor = list(setores.keys())[0]

maior_setor = primeiro_setor
menor_setor = primeiro_setor

maior_quantidade = setores[primeiro_setor]
menor_quantidade = setores[primeiro_setor]

for setor, quantidade in setores.items():

    if quantidade > maior_quantidade:
        maior_quantidade = quantidade
        maior_setor = setor

    if quantidade < menor_quantidade:
        menor_quantidade = quantidade
        menor_setor = setor

print()
print(f"Setor com mais funcionários: {maior_setor} ({maior_quantidade})")
print(f"Setor com menos funcionários: {menor_setor} ({menor_quantidade})")


# =============================
# 3. CALCULO CARGO
print("")
print("=" * 40)
print("3. CALCULO CARGO")
print("=" * 40)

# calculo quantidade cargo:
cargos = {}
cargo_destaque = ""

for funcionario in funcionarios:

    cargo = funcionario["cargo"]

    if cargo not in cargos:
        cargos[cargo] = 1

    else:
        cargos[cargo] += 1

for cargo, quantidade in cargos.items():
    print(f"{cargo} ..... {quantidade} funcionário(s)")


# calculando cargo com mais e menos funcionários
primeiro_cargo = list(cargos.keys())[0]
maior_cargo = primeiro_cargo
menor_cargo = primeiro_cargo

maior_quantidade = cargos[primeiro_cargo]
menor_quantidade = cargos[primeiro_cargo]

for cargo, quantidade in cargos.items():

    if quantidade > maior_quantidade:
        maior_quantidade = quantidade
        maior_cargo = cargo

    if quantidade < menor_quantidade:
        menor_quantidade = quantidade
        menor_cargo = cargo


print()
print(f"Cargo com mais funcionários: {maior_cargo} ({maior_quantidade})")
print(f"Cargo com menos funcionários: {menor_cargo} ({menor_quantidade})")
```

## Desafio 2 - Cadastro de E-mails
Arquivo: emails_corporativos.json

Faça:
* Ler o JSON.
* Validar todos os e-mails.
* Contar válidos.
* Contar inválidos.
* Mostrar somente os inválidos.
* Depois crie:
    * def validar_email(email):
    * e utilize a função.

Objetivo - Treinar:
* JSON
* funções
* regex

```python
import json
import re

with open("dataset/emails_corporativos.json", encoding="utf-8") as arquivo:
    emails = json.load(arquivo)

# print(emails[0])


# =================================
# Quantidade de e-mails validos e invalidos
print("")
print("=" * 40)
print("1. QUANTIDADE E-MAILS VÁLIDOS E INVALIDOS")
print("=" * 40)

padrao = r"^[^\s@]+@[^\s@]+\.[^\s@]+$"
email_validos = 0
email_invalido = 0


for email in emails:
    email = email['email']

    if re.match(padrao, email):
        email_validos += 1
    else:
        email_invalido += 1
print(f"Quantidade E-mails Validos ..... {email_validos} e-mails")
print(f"Quantidade E-mails Invalido ..... {email_invalido} e-mails\n")


# =============================
# LISTA E-MAILS INVALIDOS
print("")
print("=" * 40)
print("2. LISTA E-MAILS INVALIDOS")
print("=" * 40)

# Lista E-mails Invalidos
padrao = r"^[^\s@]+@[^\s@]+\.[^\s@]+$"
email_validos = 0
email_invalido = 0

for email in emails:
    email = email['email']

    if re.match(padrao, email):
        email_validos += 1
    else:
        email_invalido += 1
        print(email)



# =============================
# DEF validação e-mail
print("")
print("=" * 40)
print("3. VALIDAÇÃO E-MAIL")
print("=" * 40)

def validar_email(email):
    padrao = r"^[^\s@]+@[^\s@]+\.[^\s@]+$"
    if re.match(padrao, email):
        return "E-mail Valido"
    else:
        return "E-mail Invalido"

for email in emails:
    email = email['email']
    print(f"E-mail: {email} - {validar_email(email)}")
```

## Desafio 3 - Banco de Horas
Arquivo: recursos_humanos.xlsx

Faça:
* Ler o Excel.
* Mostrar os cinco primeiros registros.
* Calcular horas trabalhadas.
* Calcular atraso.
* Classificar:
    * Pontual
    * Atrasado
    * Muito atrasado

```python
from openpyxl import load_workbook

# =========================
# Leitura planilha
planilha = load_workbook("dataset/recursos_humanos.xlsx")


# =========================
# criando as variáveis
aba_cadastro = planilha["cadastro_funcionario"]
aba_ponto = planilha["banco_horas"]
funcionarios = {}


# =========================
# print primeiras 5 linhas
print("=" * 5)
print("REGISTRO FUNCIONÁRIOS")
print("=" * 5)

for cadastro in aba_cadastro.iter_rows(min_row=1, max_row=6, values_only=True):
    print(cadastro)


# =========================
# Dicionário ID > Funcionário
for linha in aba_cadastro.iter_rows(min_row=2, values_only=True):
    id_funcionario = linha[0]
    nome = linha[1]
    
    funcionarios[id_funcionario] = nome


# =========================
# calculo horas trabalhadas por funcionário
print()
print("=" * 5)
print("HORAS TRABALHADAS POR FUNCIONÁRIO")
print("=" * 5)

for linha in aba_ponto.iter_rows(min_row=2, values_only=True):
    id_funcionario = linha[0]
    entrada = linha[4]
    saida = linha[5]
    horas_trabalhadas = (saida - entrada).seconds / 3600
    nome = funcionarios[id_funcionario]
    
    print(
        f"{nome:<20}"
        f"Horas trabalhadas: {horas_trabalhadas:.2f}h")
    

# =========================
# calculo horas de atraso por funcionário
print()
print("=" * 5)
print("HORAS ATRASO POR FUNCIONÁRIO")
print("=" * 5)

for linha in aba_ponto.iter_rows(min_row=2, values_only=True):
    id_funcionario = linha[0]
    nome = funcionarios[id_funcionario]
    
    entrada_prevista_min = 8 * 60
    entrada_real = linha[4]
    entrada_real_min = entrada_real.hour * 60 + entrada_real.minute

    atraso_minutos = entrada_real_min - entrada_prevista_min

    if atraso_minutos <= 0:
        status = "Pontual"

    elif atraso_minutos <= 10:
        status = "Atrasado"

    else:
        status = "Muito Atrasado"

    print(
        f"{nome:<20}"
        f"Atraso {atraso_minutos:>3} min"
        f" | Status: {status}"
        )
```

## Desafio 4 - Dashboard RH
Com base no exercício 3, crie um Dashboard único com as seguintes informações:
* Nome
* Cargo
* Entrada
* Saída
* Horas Trabalhadas
* Atraso
* Status

```python
from openpyxl import load_workbook

# =========================
# Leitura planilha
planilha = load_workbook("dataset/recursos_humanos.xlsx")


# =========================
# criando as variáveis
aba_cadastro = planilha["cadastro_funcionario"]
aba_ponto = planilha["banco_horas"]


# =========================
# Dicionário ID > Funcionário
funcionarios = {}

for linha in aba_cadastro.iter_rows(min_row=2, values_only=True):

    id_funcionario = linha[0]

    funcionarios[id_funcionario] = {
        "Nome": linha[1],
        "Cargo": linha[2],
        "Setor": linha[3]
    }


# =========================
# CRIAÇÃO RELATÓRIO
print("=" * 60)
print("RELATÓRIO DE BANCO DE HORAS")
print("=" * 60)

for linha in aba_ponto.iter_rows(min_row=2, values_only=True):

    id_funcionario = linha[0]

    entrada = linha[4]
    saida = linha[5]

    horas_trabalhadas = (saida - entrada).seconds / 3600

    entrada_prevista = 8 * 60
    entrada_real = entrada.hour * 60 + entrada.minute

    atraso = entrada_real - entrada_prevista

    if atraso <= 0:
        status = "Pontual"

    elif atraso <= 10:
        status = "Atrasado"

    else:
        status = "Muito Atrasado"

    nome = funcionarios[id_funcionario]["Nome"]
    cargo = funcionarios[id_funcionario]["Cargo"]
    setor = funcionarios[id_funcionario]["Setor"]

    print(f"{nome}")
    print(f"Cargo.......: {cargo}")
    print(f"Setor.......: {setor}")
    print(f"Entrada.....: {entrada.strftime('%H:%M')}")
    print(f"Saída.......: {saida.strftime('%H:%M')}")
    print(f"Horas.......: {horas_trabalhadas:.2f} h")
    print(f"Atraso......: {atraso} min")
    print(f"Status......: {status}")
    print("-" * 60)
```
