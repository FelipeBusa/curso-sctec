**from Ipython.display import display** = Serve para importar o display para o código.

**.describe()** = Serve para criar um resumo estatístico das colunas numéricas de um DataFrame
> print(vendas.describe())
>> saídas:
>>> count - Quantidade de valores preenchidos.

>>> mean - média

>>> std - É o desvio padrão. Mostra o quanto os valores variam em relação à média.

>>>> std pequeno → dados parecidos

>>>> std grande → dados muito diferentes

>>> min - Menor valor.

>>> 25% - 1º quartil. Significa que 25% dos valores são menores ou iguais a esse número.

>>> 50% - É a mediana. Muita gente confunde com média.

>>> 75% - 3º quartil. Significa que 75% dos valores são menores ou iguais a esse número.

>>> max - Maior valor.

**.info()** = Serve para visualizar um resumo das informações, mostrando o tipo da coluna e quantidade de informações preenchidas (avaliar casos vazios)

**.groupby()** = Serve para agruoar conforme regra estabelecida.
