### Desafio 1 Aula:
Vendas mensais de uma loja de roupas. Arquivo: dataset_squad_b_loja_roupas.csv

Faça uma análise das vendas por categoria e mês .
**Comando:** rode `head()`, `describe()` e pelo menos um `groupby()` e apresente 2 conclusões reais que ajudem a diretoria a decidir o mix de compras.

```python
import pandas as pd
from IPython.display import display

dataset_squad_b = pd.read_csv("dataset/dataset_squad_b_loja_roupas.csv")

display(dataset_squad_b.head())
display(dataset_squad_b.describe())
display(dataset_squad_b.groupby("categoria")["vendas"].sum().sort_values(ascending=False))
display(dataset_squad_b.groupby("mes")["vendas"].sum())

# Conclusão 1: "Camisas" é a categoria mais vendida no período (760 unidades), quase o dobro de "Casacos" (320) — vale reforçar estoque de camisas.
# Conclusão 2: Março foi o mês mais fraco (320 unidades) e Fevereiro o mais forte (400) — vale investigar o que mudou em março (promoção? sazonalidade?).
```

### Desafio 2 Aula:
Assinantes de streaming por região e plano. Arquivo: dataset_squad_c_streaming.csv

Faça um mapeamento de onde estão os assinantes hoje e quais planos vendem mais em cada região.
**Comando:** rode `head()`, `describe()` e pelo menos um `groupby()` e apresente 2 conclusões reais que ajudem a decidir onde investir.


```python
import pandas as pd
from IPython.display import display

dataset_squad_c = pd.read_csv("dataset/dataset_squad_c_streaming.csv")

display(dataset_squad_c.head())
display(dataset_squad_c.describe())
display(dataset_squad_c.groupby("regiao")["assinantes"].sum().sort_values(ascending=False))
display(dataset_squad_c.groupby("plano")["assinantes"].sum().sort_values(ascending=False))

# Conclusão 1: a região Sudeste concentra quase 40% de todos os assinantes (28.200 de ~72.100) — a região Norte tem só 2.900, o menor volume.
# Conclusão 2: o plano Básico é o mais assinado (25.400), à frente de Premium (18.300) e Família (11.900) — sugere que preço pesa mais que recursos extras para a maioria dos assinantes.
```

### Desafio 3 Aula
Avaliações de um app. Arquivo: dataset_squad_d_app.csv

Faça uma análise para saber quais categorias de app têm melhor avaliação e mais downloads.
**Comando:** rode `head()`, `describe()` e pelo menos um `groupby()` e apresente 2 conclusões reais que ajudem a escolher a categoria do próximo lançamento.

```python
import pandas as pd
from IPython.display import display

dataset_squad_d = pd.read_csv("dataset/dataset_squad_d_app.csv")

display(dataset_squad_d.head())
display(dataset_squad_d.describe())
display(dataset_squad_d.groupby("categoria_app")[["nota", "downloads_mil"]].mean().sort_values("nota", ascending=False))

# Conclusão 1: "Educação" tem a melhor nota média (4.65), mas é a categoria com MENOS downloads (81 mil) — nota alta não garante volume.
# Conclusão 2: "Jogos" tem a pior nota média (4.03) e, ainda assim, de longe o maior volume de downloads (532 mil) — o próximo lançamento precisa decidir entre "nota" e "alcance" como prioridade.
```

### Desafio 4
Atendimentos de uma clínica. Arquivo: dataset_squad_e_clinica.csv

Faça uma análise para entender se o tempo de espera está afetando a satisfação dos pacientes, especialidade por especialidade, antes de decidir se contrata mais profissionais.

**Comando:** rode `head()`, `describe()` e pelo menos um `groupby()` e apresente 2 conclusões reais que ajudem a decidir sobre a contratação.

```python
import pandas as pd
from IPython.display import display

dataset_squad_e = pd.read_csv("dataset/dataset_squad_e_clinica.csv")

display(dataset_squad_e.head())
display(dataset_squad_e.describe())
display(dataset_squad_e.groupby("especialidade")[["tempo_espera_min", "satisfacao"]].mean().sort_values("tempo_espera_min", ascending=False))

# Conclusão 1: Ortopedia tem o maior tempo médio de espera (39 min) e a menor satisfação média (3.2) entre as especialidades.
# Conclusão 2: Pediatria tem o menor tempo de espera (18.75 min) e a maior satisfação (4.75) — os dados sugerem que tempo de espera e satisfação andam juntos, o que apoiaria reforçar equipe onde a espera é maior.
```
