# 🚀 Projeto 1 - Sistema Comercial Completo
Objetivo: Uma empresa deseja gerar um relatório completo de vendas.
1. Mostre todas as vendas.
2. Calcule
    * valor total vendido
    * quantidade total de itens vendidos
    * ticket médio por venda
    * maior venda
    * menor venda
3. Conte quantas vendas cada vendedor realizou
4. Calcule quanto cada vendedor faturou.
5. Calcule a média das vendas.
6. Classifique os vendedores:
    * Mais de R$10000 = Excelente
    * Mais de R$7000 = Bom
    * Caso contrário = Regular
7. Descubra:
    * vendedor que mais vendeu
    * vendedor que menos vendeu
8. Agrupe por cidade.
9. Monte um resumo final:
    * TOTAL VENDIDO
    * MAIOR VENDA
    * MENOR VENDA
    * MÉDIA
    * ITENS VENDIDOS
    * VENDEDOR DESTAQUE
    * CIDADE DESTAQUE

```python
vendas = [
    {"Vendedor":"Ana","Cidade":"Curitiba","Produto":"Notebook","Quantidade":2,"Valor":3500},
    {"Vendedor":"Carlos","Cidade":"Joinville","Produto":"Mouse","Quantidade":1,"Valor":80},
    {"Vendedor":"Maria","Cidade":"Curitiba","Produto":"Monitor","Quantidade":3,"Valor":1200},
    {"Vendedor":"Ana","Cidade":"Curitiba","Produto":"Teclado","Quantidade":5,"Valor":150},
    {"Vendedor":"Carlos","Cidade":"Joinville","Produto":"Notebook","Quantidade":1,"Valor":3500},
    {"Vendedor":"João","Cidade":"Blumenau","Produto":"Mouse","Quantidade":15,"Valor":80},
    {"Vendedor":"Maria","Cidade":"Curitiba","Produto":"Notebook","Quantidade":2,"Valor":3500},
    {"Vendedor":"Ana","Cidade":"Curitiba","Produto":"Monitor","Quantidade":1,"Valor":1200}
]

# Váriaveis
venda_total = 0
quantidade_total = 0
ticket_medio = 0
primeira_venda = vendas[0]["Quantidade"] * vendas[0]["Valor"]
maior_venda = primeira_venda
menor_venda = primeira_venda
maior_vendedor = vendas[0]["Vendedor"]
menor_vendedor = vendas[0]["Vendedor"]
total_vendedor = {}
total_cidade = {}
cidade_destaque = vendas[0]["Cidade"]

# ==========================
# 1. RELATÓRIO DAS VENDAS
# for calculando valores individuais por vendedor/venda
print("=" * 40)
print(" 1. RELATÓRIO DAS VENDAS")
print("=" * 40)
for venda in vendas:
  vendedor = venda["Vendedor"]
  cidade = venda["Cidade"]
  produto = venda["Produto"]
  quantidade = venda["Quantidade"]
  valor = venda["Valor"]
  valor_total_venda  = valor * quantidade


  print(f"Vendedor(a): {vendedor}")
  print(f"Cidade: {cidade}")
  print(f"Produto: {produto}")
  print(f"Quantidade: {quantidade}")
  print(f"Valor Produto: R$ {valor:.2f}")
  print(f"Valor Total da Venda: R$ {valor_total_venda:.2f}\n")


# ==========================
# 2. INDICADORES GERAIS
# for calculando valores gerais
print()
print("=" * 40)
print("2. INDICADORES GERAIS")
print("=" * 40)

for venda in vendas:
  vendedor = venda["Vendedor"]
  quantidade = venda["Quantidade"]
  valor = venda["Valor"]
  valor_total_venda  = valor * quantidade
  venda_total += valor_total_venda
  quantidade_total += quantidade


  if valor_total_venda > maior_venda:
    maior_venda = valor_total_venda
    maior_vendedor = vendedor

  if valor_total_venda < menor_venda:
    menor_venda = valor_total_venda
    menor_vendedor = vendedor

ticket_medio = venda_total / len(vendas)
media_por_item_vendido = venda_total/quantidade_total

print(f"Venda Total ..... R$ {venda_total:.2f}")
print(f"Quantidade Total de Produtos ..... {quantidade_total}")
print(f"Quantidade de Vendas ..... {len(vendas)}\n")

print(f"Ticket Médio ..... R$ {ticket_medio:.2f}")
print(f"Valor Médio Por Produto ..... R$ {media_por_item_vendido:.2f}\n")

print(f"Maior Venda ..... {maior_vendedor} ..... R$ {maior_venda:.2f}")
print(f"Menor Venda ..... {menor_vendedor} ..... R$ {menor_venda:.2f}\n")


# ==========================
# 3. RESUMO POR VENDEDOR
# for resumo valores individuais por vendedor
print()
print("=" * 40)
print("3. RESUMO POR VENDEDOR")
print("=" * 40)

for venda in vendas:
  vendedor = venda["Vendedor"]
  quantidade = venda["Quantidade"]
  valor = venda["Valor"]

  valor_total_venda  = valor * quantidade
  if vendedor not in total_vendedor:
    total_vendedor[vendedor] = {
        "Quantidade_Vendas": 1,
        "Total_Vendido": valor_total_venda
    }

  else:
    total_vendedor[vendedor]["Quantidade_Vendas"] += 1
    total_vendedor[vendedor]["Total_Vendido"] += valor_total_venda

for vendedor, dados in total_vendedor.items():
    print(f"Vendedor: {vendedor}")
    print(f"Quantidade de vendas: {dados['Quantidade_Vendas']}")
    print(f"Total vendido: R$ {dados['Total_Vendido']:.2f}\n")


# ==========================
# 4. CLASSIFICAÇÃO DOS VENDEDORES
# for para classificar os vendedores (excelente, bom e regular)
print()
print("=" * 40)
print("4. CLASSIFICAÇÃO DOS VENDEDORES")
print("=" * 40)

for vendedor, dados in total_vendedor.items():
  if dados['Total_Vendido'] > 10000:
    status = "Excelente"
  elif dados['Total_Vendido'] > 7000:
    status = "Bom"
  else:
    status = "Regular"

  print(f"Vendedor: {vendedor}")
  print(f"Classificação: {status}\n")


# ==========================
# 5. RESUMO POR CIDADE
print()
print("=" * 40)
print("5. RESUMO POR CIDADE")
print("=" * 40)

for venda in vendas:
  cidade = venda["Cidade"]
  quantidade = venda["Quantidade"]
  valor = venda["Valor"]
  valor_total_venda  = valor * quantidade

  if cidade not in total_cidade:
    total_cidade[cidade] = {
        "Quantidade_Vendas": 1,
        "Total_Vendido": valor_total_venda
    }

  else:
    total_cidade[cidade]["Quantidade_Vendas"] += 1
    total_cidade[cidade]["Total_Vendido"] += valor_total_venda

for cidade, dados in total_cidade.items():
    print(f"Cidade: {cidade}")
    print(f"Quantidade de vendas: {dados['Quantidade_Vendas']}")
    print(f"Total vendido: R$ {dados['Total_Vendido']:.2f}\n")


# ==========================
# 6. VENDEDOR E CIDADE DESTAQUE
print()
print("=" * 40)
print("6. VENDEDOR DESTAQUE")
print("=" * 40)

vendedor_destaque = ""
maior_total_vendido = 0
cidade_destaque = ""
maior_total_cidade = 0

for vendedor, dados in total_vendedor.items():
    total_vendido = dados["Total_Vendido"]

    if total_vendido > maior_total_vendido:
        maior_total_vendido = total_vendido
        vendedor_destaque = vendedor

print(f"Vendedor Destaque: {vendedor_destaque}")
print(f"Total Vendido: R$ {maior_total_vendido:.2f}")

for cidade, dados in total_cidade.items():

    total_vendido = dados["Total_Vendido"]

    if total_vendido > maior_total_cidade:
        maior_total_cidade = total_vendido
        cidade_destaque = cidade

print(f"\nCidade Destaque: {cidade_destaque}")
print(f"Total Vendido: R$ {maior_total_cidade:.2f}")
```
