# 🚀 Projeto 1 - Sistema Comercial Completo
Objetivo: Uma empresa deseja gerar um relatório completo de vendas.
1. Mostre todas as vendas.
2. Calcule
    * valor total vendido
    * quantidade total de itens vendidos
    * ticket médio por venda
    * maior venda
    * menor venda
3. Conte quantas vendas cada vendedor realizou
4. Calcule quanto cada vendedor faturou.
5. Calcule a média das vendas.
6. Classifique os vendedores:
    * Mais de R$10000 = Excelente
    * Mais de R$7000 = Bom
    * Caso contrário = Regular
7. Descubra:
    * vendedor que mais vendeu
    * vendedor que menos vendeu
8. Agrupe por cidade.
9. Monte um resumo final:
    * TOTAL VENDIDO
    * MAIOR VENDA
    * MENOR VENDA
    * MÉDIA
    * ITENS VENDIDOS
    * VENDEDOR DESTAQUE
    * CIDADE DESTAQUE

```python
vendas = [
    {"Vendedor":"Ana","Cidade":"Curitiba","Produto":"Notebook","Quantidade":2,"Valor":3500},
    {"Vendedor":"Carlos","Cidade":"Joinville","Produto":"Mouse","Quantidade":1,"Valor":80},
    {"Vendedor":"Maria","Cidade":"Curitiba","Produto":"Monitor","Quantidade":3,"Valor":1200},
    {"Vendedor":"Ana","Cidade":"Curitiba","Produto":"Teclado","Quantidade":5,"Valor":150},
    {"Vendedor":"Carlos","Cidade":"Joinville","Produto":"Notebook","Quantidade":1,"Valor":3500},
    {"Vendedor":"João","Cidade":"Blumenau","Produto":"Mouse","Quantidade":15,"Valor":80},
    {"Vendedor":"Maria","Cidade":"Curitiba","Produto":"Notebook","Quantidade":2,"Valor":3500},
    {"Vendedor":"Ana","Cidade":"Curitiba","Produto":"Monitor","Quantidade":1,"Valor":1200}
]

# Váriaveis
venda_total = 0
quantidade_total = 0
ticket_medio = 0
primeira_venda = vendas[0]["Quantidade"] * vendas[0]["Valor"]
maior_venda = primeira_venda
menor_venda = primeira_venda
maior_vendedor = vendas[0]["Vendedor"]
menor_vendedor = vendas[0]["Vendedor"]
total_vendedor = {}
total_cidade = {}
cidade_destaque = vendas[0]["Cidade"]

# ==========================
# 1. RELATÓRIO DAS VENDAS
# for calculando valores individuais por vendedor/venda
print("=" * 40)
print(" 1. RELATÓRIO DAS VENDAS")
print("=" * 40)
for venda in vendas:
  vendedor = venda["Vendedor"]
  cidade = venda["Cidade"]
  produto = venda["Produto"]
  quantidade = venda["Quantidade"]
  valor = venda["Valor"]
  valor_total_venda  = valor * quantidade


  print(f"Vendedor(a): {vendedor}")
  print(f"Cidade: {cidade}")
  print(f"Produto: {produto}")
  print(f"Quantidade: {quantidade}")
  print(f"Valor Produto: R$ {valor:.2f}")
  print(f"Valor Total da Venda: R$ {valor_total_venda:.2f}\n")


# ==========================
# 2. INDICADORES GERAIS
# for calculando valores gerais
print()
print("=" * 40)
print("2. INDICADORES GERAIS")
print("=" * 40)

for venda in vendas:
  vendedor = venda["Vendedor"]
  quantidade = venda["Quantidade"]
  valor = venda["Valor"]
  valor_total_venda  = valor * quantidade
  venda_total += valor_total_venda
  quantidade_total += quantidade


  if valor_total_venda > maior_venda:
    maior_venda = valor_total_venda
    maior_vendedor = vendedor

  if valor_total_venda < menor_venda:
    menor_venda = valor_total_venda
    menor_vendedor = vendedor

ticket_medio = venda_total / len(vendas)
media_por_item_vendido = venda_total/quantidade_total

print(f"Venda Total ..... R$ {venda_total:.2f}")
print(f"Quantidade Total de Produtos ..... {quantidade_total}")
print(f"Quantidade de Vendas ..... {len(vendas)}\n")

print(f"Ticket Médio ..... R$ {ticket_medio:.2f}")
print(f"Valor Médio Por Produto ..... R$ {media_por_item_vendido:.2f}\n")

print(f"Maior Venda ..... {maior_vendedor} ..... R$ {maior_venda:.2f}")
print(f"Menor Venda ..... {menor_vendedor} ..... R$ {menor_venda:.2f}\n")


# ==========================
# 3. RESUMO POR VENDEDOR
# for resumo valores individuais por vendedor
print()
print("=" * 40)
print("3. RESUMO POR VENDEDOR")
print("=" * 40)

for venda in vendas:
  vendedor = venda["Vendedor"]
  quantidade = venda["Quantidade"]
  valor = venda["Valor"]

  valor_total_venda  = valor * quantidade
  if vendedor not in total_vendedor:
    total_vendedor[vendedor] = {
        "Quantidade_Vendas": 1,
        "Total_Vendido": valor_total_venda
    }

  else:
    total_vendedor[vendedor]["Quantidade_Vendas"] += 1
    total_vendedor[vendedor]["Total_Vendido"] += valor_total_venda

for vendedor, dados in total_vendedor.items():
    print(f"Vendedor: {vendedor}")
    print(f"Quantidade de vendas: {dados['Quantidade_Vendas']}")
    print(f"Total vendido: R$ {dados['Total_Vendido']:.2f}\n")


# ==========================
# 4. CLASSIFICAÇÃO DOS VENDEDORES
# for para classificar os vendedores (excelente, bom e regular)
print()
print("=" * 40)
print("4. CLASSIFICAÇÃO DOS VENDEDORES")
print("=" * 40)

for vendedor, dados in total_vendedor.items():
  if dados['Total_Vendido'] > 10000:
    status = "Excelente"
  elif dados['Total_Vendido'] > 7000:
    status = "Bom"
  else:
    status = "Regular"

  print(f"Vendedor: {vendedor}")
  print(f"Classificação: {status}\n")


# ==========================
# 5. RESUMO POR CIDADE
print()
print("=" * 40)
print("5. RESUMO POR CIDADE")
print("=" * 40)

for venda in vendas:
  cidade = venda["Cidade"]
  quantidade = venda["Quantidade"]
  valor = venda["Valor"]
  valor_total_venda  = valor * quantidade

  if cidade not in total_cidade:
    total_cidade[cidade] = {
        "Quantidade_Vendas": 1,
        "Total_Vendido": valor_total_venda
    }

  else:
    total_cidade[cidade]["Quantidade_Vendas"] += 1
    total_cidade[cidade]["Total_Vendido"] += valor_total_venda

for cidade, dados in total_cidade.items():
    print(f"Cidade: {cidade}")
    print(f"Quantidade de vendas: {dados['Quantidade_Vendas']}")
    print(f"Total vendido: R$ {dados['Total_Vendido']:.2f}\n")


# ==========================
# 6. VENDEDOR E CIDADE DESTAQUE
print()
print("=" * 40)
print("6. VENDEDOR DESTAQUE")
print("=" * 40)

vendedor_destaque = ""
maior_total_vendido = 0
cidade_destaque = ""
maior_total_cidade = 0

for vendedor, dados in total_vendedor.items():
    total_vendido = dados["Total_Vendido"]

    if total_vendido > maior_total_vendido:
        maior_total_vendido = total_vendido
        vendedor_destaque = vendedor

print(f"Vendedor Destaque: {vendedor_destaque}")
print(f"Total Vendido: R$ {maior_total_vendido:.2f}")

for cidade, dados in total_cidade.items():

    total_vendido = dados["Total_Vendido"]

    if total_vendido > maior_total_cidade:
        maior_total_cidade = total_vendido
        cidade_destaque = cidade

print(f"\nCidade Destaque: {cidade_destaque}")
print(f"Total Vendido: R$ {maior_total_cidade:.2f}")
```

# 🚀 Projeto 3 — Dashboard de Recursos Humanos
Contexto: Uma empresa possui o cadastro dos seus colaboradores e deseja gerar um relatório completo para auxiliar o setor de RH.

1. Relatório Geral. Mostre para cada funcionário:
    * Nome
    * Cargo
    * Departamento
    * Salário
    * Tempo de empresa

2. Classificação Salarial. Classifique cada funcionário:
    * Salário acima de 10.000 = Executivo
    * Entre 6.000 e 10.000 = Alta Gestão
    * Entre 4.000 e 5.999 = Especialista
    * Abaixo de 4.000 = Operacional

        - Mostre:
            - Maria
            - Salário: R$5200 
            - Classificação: Especialista

3. Tempo de Empresa. Classifique:
    * Até 2 anos = Novo
    * 3 até 5 anos = Experiente
    * Acima de 5 anos = Veterano

4. Estatísticas Gerais. Calcule:
    * Total da folha salarial
    * Média salarial
    * Maior salário
    * Menor salário
    * Quantidade de funcionários

5. Acima e Abaixo da Média. Depois de calcular a média salarial. Conte:
    * Quantos funcionários recebem acima da média.
    * Quantos recebem abaixo da média.
    * Quantos recebem exatamente a média (se existir).

6. Departamento. Conte quantos funcionários existem em cada departamento.

7. Gasto por Departamento. Calcule quanto a empresa gasta com salários em cada departamento.

8. Funcionário Destaque. Descubra:
    * funcionário com maior salário
    * funcionário com menor salário
    * funcionário mais antigo
    * funcionário mais novo

9. Dashboard Final. Monte um relatório organizado. Exemplo:

_______
        DASHBOARD RH
Funcionários ............ 8

Folha Salarial .......... R$ xx.xxx

Média Salarial .......... R$ x.xxx

Maior Salário ........... R$ x.xxx

Menor Salário ........... R$ x.xxx

Acima da Média .......... x

Abaixo da Média ......... x

Maior Salário ........... Patrícia

Menor Salário ........... João

Mais Antigo ............. Patrícia

Mais Novo ............... Bruno

```python
funcionarios = [
    {"Nome":"Ana","Cargo":"Analista","Salario":4500,"Tempo":3,"Departamento":"Comercial"},
    {"Nome":"Carlos","Cargo":"Gerente","Salario":8500,"Tempo":8,"Departamento":"Financeiro"},
    {"Nome":"Maria","Cargo":"Analista","Salario":5200,"Tempo":5,"Departamento":"RH"},
    {"Nome":"João","Cargo":"Assistente","Salario":2800,"Tempo":2,"Departamento":"Comercial"},
    {"Nome":"Fernanda","Cargo":"Coordenador","Salario":6700,"Tempo":6,"Departamento":"TI"},
    {"Nome":"Lucas","Cargo":"Analista","Salario":4800,"Tempo":4,"Departamento":"TI"},
    {"Nome":"Patrícia","Cargo":"Diretor","Salario":15000,"Tempo":12,"Departamento":"Diretoria"},
    {"Nome":"Bruno","Cargo":"Assistente","Salario":3000,"Tempo":1,"Departamento":"RH"}
]

# ========================
# 1. RELATÓRIO GERAL

print("=" * 40)
print("1. RELATÓRIO GERAL")
print("=" * 40)

for funcionario in funcionarios:
  nome_funcionario = funcionario['Nome']
  cargo = funcionario['Cargo']
  departamento = funcionario['Departamento']
  salario_funcionario = f"{funcionario['Salario']:,.2f}".replace(",", "X").replace(".", ",").replace("X", ".")
  tempo_funcionario = funcionario['Tempo']

  print(f"Funcionário(a): {nome_funcionario}")
  print(f"Cargo: {cargo}")
  print(f"Departamento: {departamento}")
  print(f"Salário: R$ {salario_funcionario}")
  print(f"Tempo de Empresa: {tempo_funcionario} anos\n")


# ========================
# 2. CLASSIFICAÇÃO SALARIAL
print()
print("=" * 40)
print("2. CLASSIFICAÇÃO SALARIAL")
print("=" * 40)

for funcionario in funcionarios:
  nome_funcionario = funcionario['Nome']
  cargo = funcionario['Cargo']
  salario_funcionario = funcionario['Salario']

  if salario_funcionario > 10000:
    classificacao_salarial_funcionario = "Executivo"
  elif salario_funcionario > 6000:
    classificacao_salarial_funcionario = "Alta Gestão"
  elif salario_funcionario > 4000:
    classificacao_salarial_funcionario = "Especialista"
  else:
    classificacao_salarial_funcionario = "Operacional"

  print(f"Funcionário(a): {nome_funcionario}")
  print(f"Salário: R$ {funcionario['Salario']:,.2f}".replace(",", "X").replace(".", ",").replace("X", "."))
  print(f"Classificação Salarial: {classificacao_salarial_funcionario}\n")


# ========================
# 3. TEMPO DE EMPRESA
print()
print("=" * 40)
print("3. TEMPO DE EMPRESA")
print("=" * 40)

for funcionario in funcionarios:
  nome_funcionario = funcionario['Nome']
  tempo_funcionario = funcionario['Tempo']

  if tempo_funcionario > 5:
    tempo_casa = "Veterano"
  elif tempo_funcionario >= 3:
    tempo_casa = "Experiente"
  else:
    tempo_casa = "Novo"

  print(f"Funcionário(a): {nome_funcionario}")
  print(f"Tempo de Empresa: {tempo_casa} ..... {tempo_funcionario} ano(s).\n")


# ========================
# 4. ESTATÍSTICAS GERAIS
print()
print("=" * 40)
print("4. ESTATÍSTICAS GERAIS")
print("=" * 40)

salario_total = 0


for funcionario in funcionarios:
  nome_funcionario = funcionario['Nome']
  cargo = funcionario['Cargo']
  salario_funcionario = funcionario['Salario']
  salario_total += salario_funcionario

media_salarial = salario_total / len(funcionarios)
print(f"Total Folha Salarial: R$ {salario_total:,.2f}".replace(",","X").replace(".",",").replace("X","."))
print(f"Média Salaria: R$ {media_salarial:,.2f}".replace(",","X").replace(".",",").replace("X","."))



salario_maximo = funcionarios[0]["Salario"]
salario_minimo = funcionarios[0]["Salario"]
nome_maior_salario = funcionarios[0]["Nome"]
nome_menor_salario = funcionarios[0]["Nome"]

for funcionario in funcionarios:

    nome = funcionario["Nome"]
    salario = funcionario["Salario"]

    if salario > salario_maximo:
        salario_maximo = salario
        nome_maior_salario = nome

    if salario < salario_minimo:
        salario_minimo = salario
        nome_menor_salario = nome

print()
print(f"Maior salário: {nome_maior_salario} - R$ {salario_maximo:.2f}")
print(f"Menor salário: {nome_menor_salario} - R$ {salario_minimo:.2f}")


qtd_baixo_media = 0
qtd_alto_media = 0
qtd_na_media = 0

for funcionario in funcionarios:
  salario_funcionario = funcionario['Salario']
  salario_total += salario_funcionario
  media_salarial = salario_total / len(funcionarios)

  if salario_funcionario > media_salarial:
    qtd_alto_media += 1
  elif salario_funcionario < media_salarial:
    qtd_baixo_media += 1
  else:
    qtd_na_media += 1

print()
print(f"Quantidade de funcionários com salário ACIMA da média: {qtd_alto_media} funcionário(s)")
print(f"Quantidade de funcionários com salário na média: {qtd_na_media} funcionário(s)")
print(f"Quantidade de funcionários com salário ABAIXO da média: {qtd_baixo_media} funcionário(s)")



# ========================
# 5.FUNCIONÁRIOS POR DEPARTAMENTO
print()
print("=" * 40)
print("5. FUNCIONÁRIOS POR DEPARTAMENTO")
print("=" * 40)

departamentos = {}

for funcionario in funcionarios:

    departamento = funcionario["Departamento"]

    if departamento not in departamentos:
        departamentos[departamento] = 1

    else:
        departamentos[departamento] += 1

for departamento, quantidade in departamentos.items():
    print(f"{departamento} ..... {quantidade} funcionário(s)")


# ==========================
# 6. GASTO POR DEPARTAMENTO
print()
print("=" * 40)
print("6. GASTO POR DEPARTAMENTO")
print("=" * 40)

gasto_departamento = {}

for funcionario in funcionarios:

    departamento = funcionario["Departamento"]
    salario = funcionario["Salario"]

    if departamento not in gasto_departamento:
        gasto_departamento[departamento] = salario

    else:
        gasto_departamento[departamento] += salario

for departamento, total in gasto_departamento.items():
    print(f"{departamento} ..... R$ {total:.2f}")


# ==========================
# 7.FUNCIONÁRIO DESTAQUE
print()
print("=" * 40)
print("7. FUNCIONÁRIOS DESTAQUE")
print("=" * 40)

# Valores iniciais
maior_salario = funcionarios[0]["Salario"]
menor_salario = funcionarios[0]["Salario"]

mais_antigo = funcionarios[0]["Tempo"]
mais_novo = funcionarios[0]["Tempo"]

nome_maior_salario = funcionarios[0]["Nome"]
nome_menor_salario = funcionarios[0]["Nome"]

nome_mais_antigo = funcionarios[0]["Nome"]
nome_mais_novo = funcionarios[0]["Nome"]

for funcionario in funcionarios:

    nome = funcionario["Nome"]
    salario = funcionario["Salario"]
    tempo = funcionario["Tempo"]

    # Maior salário
    if salario > maior_salario:
        maior_salario = salario
        nome_maior_salario = nome

    # Menor salário
    if salario < menor_salario:
        menor_salario = salario
        nome_menor_salario = nome

    # Mais antigo
    if tempo > mais_antigo:
        mais_antigo = tempo
        nome_mais_antigo = nome

    # Mais novo
    if tempo < mais_novo:
        mais_novo = tempo
        nome_mais_novo = nome

print(f"Maior salário ..... {nome_maior_salario} - R$ {maior_salario:.2f}")
print(f"Menor salário ..... {nome_menor_salario} - R$ {menor_salario:.2f}")

print()

print(f"Funcionário mais antigo ..... {nome_mais_antigo} - {mais_antigo} anos")
print(f"Funcionário mais novo ..... {nome_mais_novo} - {mais_novo} ano(s)")
```
