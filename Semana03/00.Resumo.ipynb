## LISTA
Sequência dentro de colchetes [ ], separados por vírgulas:

```python
frutas = ["maçã", "banana", "laranja"]
notas  = [8.5, 7.0, 9.5]
misto  = ["Ana", 25, True]
```

### Acessar os itens
Deve usar posição numérica:

`frutas[0] == "maçã"`

`frutas[-1] == "laranja"`

### Principais funções e métodos

| Recurso | O que faz | Exemplo | Resultado |
|---|---|---|---|
| `.append(valor)` | Adiciona um item ao **final** da lista | `frutas.append("uva")` | lista cresce 1 item |
| `.insert(índice, valor)` | Insere um item numa **posição específica** | `frutas.insert(1, "uva")` | "uva" entra na posição 1 |
| `.remove(valor)` | Remove a **primeira ocorrência** do valor | `frutas.remove("banana")` | só essa ocorrência some |
| `.pop(índice)` | Remove e **retorna** o item da posição (ou o último, sem índice) | `frutas.pop()` | retorna e remove o último item |
| `.sort()` | Ordena a lista em **ordem crescente** | `notas.sort()` | lista fica ordenada |
| `len(lista)` | Retorna a **quantidade** de itens | `len(frutas)` | `3` |
| `max(lista)` | Retorna o **maior** valor | `max(notas)` | `9.5` |
| `min(lista)` | Retorna o **menor** valor | `min(notas)` | `7.0` |

### Exemplo 1 — Lista de compras

```python
compras = ["leite", "arroz", "feijão"]

print(compras[0])
print(compras[-1])

# .append()

compras.append("café")
print(compras)

# .insert()

compras.insert(0, "manteiga")
print(compras)

# .remove()

compras.remove("feijão")
print(compras)

# .pop

item_retirado = compras.pop()
print(f"Retirado da lista foi: {item_retirado}")
print(compras)

# .sort

compras.sort()
print(compras)
```

### Exemplo 2 — Calculando a média de uma lista de notas

```python
notas = [7.5, 8.0, 6.5, 9.0, 5.5]

media = sum(notas) / len(notas)
print(f"Média da turma é: {media}")
```

### Exemplo 1 — Cadastro de produto

```python
produto = {
    "nome": "Mouse",
    "preco": 45.0,
    "estoque": 12
}

print(f"{produto['nome']} custa R$ {produto['preco']}")
print(f"Estoque disponível é: {produto['estoque']} unidades")
```

### Exemplo 2 — Atualizando um valor no dicionário

```python
aluno = {
    "nome": "Carlos", 
    "nota": 7.5
    }
print(f"A nota atual é: {aluno['nota']}")

aluno["nota"] = 9.0
print(f"Nota corrigida: {aluno['nota']}")
```

## LISTA DE DICIONÁRIOS


```python
carrinho = [
    {"produto": "caderno", "preco": 12.5, "quantidade": 2},
    {"produto": "lapis", "preco": 2.0, "quantidade": 5},
    {"produto": "borracha", "preco": 1.5, "quantidade": 3},
]

total = 0

for item in carrinho:
    subtotal = item["preco"] * item["quantidade"]
    total = total + subtotal
    print(f"{item['produto']}: {item['quantidade']} x R$ {item['preco']} = R$ {subtotal}")

print(f"\nTotal do carrinho é: R$ {total}")
```
