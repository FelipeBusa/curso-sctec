# RESUMO SEMANA 03
Versionamento (Git/GitHub) e Python Local

* **git clone https....** = Clona o repositório
* **cd nomedapasta** = abri pasta do repositório
* **cd ..** = volta uma pasta do repositório
* **code .** = Abre uma nova janela do VSCode
---
* **git pull** = Atualiza o seu repositório com as novas informações do GitHub
* **git status** = Mostra os arquivos que tiveram moditificação
* **git add** = Obrigatório informar arquivo ou pasta que será atualizado no repositório
* **git add .** = atualiza todos os documentos desse repositório
* **git commit -m "texto explicando o que é"** = Serve para explicar o motivo dessa atualização
* **git push -u origin main** = envia o arquivo para o GitHub na primeira vez
* **git push** = envia o arquivo para o GitHub nas próximas oportunidades
---
* **mkdir** = Cria pasta no diretório.
* **ni main.** = Cria um arquivo no diretório. Pode ser `.py = documento python` ou `.ipynb = documento notebook jupyter`
